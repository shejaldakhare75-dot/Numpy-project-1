## 1. Import NumPy and create the salary array

NumPy is imported with the alias `np`. The sample salary values are stored in a NumPy array for numerical calculations.

````python
import numpy as np

salaries = np.array([70, 65, 92, 56, 235, 250, 85, 89, 100, 75, 35, 45, 55, 30, 90])
print("Employee salaries:", salaries)
````

## 2. Basic salary statistics

These functions calculate the number of employees and key measures of central tendency.

````python
print("Total Employees:", salaries.size)
print("Maximum Salary:", np.max(salaries))
print("Minimum Salary:", np.min(salaries))
print("Average Salary:", np.mean(salaries))
print("Median Salary:", np.median(salaries))
````

## 3. Standard deviation and variance

Standard deviation and variance describe how spread out the salary values are. NumPy calculates population standard deviation and population variance by default (`ddof=0`).

````python
standard_deviation = np.std(salaries)
print("Standard Deviation:", standard_deviation)

variance = np.var(salaries)
print("Variance:", variance)
````

## 4. Filter salaries

Boolean indexing selects values that satisfy a condition. Here, salaries greater than 90 are placed in the high-salary group, while salaries below 90 are placed in the low-salary group. A salary equal to 90 is not included in either group.

````python
high_sal = salaries[salaries > 90]
print("High Salaries (above 90):", high_sal)

low_sal = salaries[salaries < 90]
print("Low Salaries (below 90):", low_sal)
````

## 5. Project salary growth over five years

This calculation assumes every salary increases by 5% each year and the increase is compounded annually. The formula is `current salary × (1.05 ** 5)`.

````python
# Estimate salaries after five years of 5% annual growth
future_sal = salaries * (1.05 ** 5)
print("Salaries after 5 years:", np.round(future_sal, 2))
````
